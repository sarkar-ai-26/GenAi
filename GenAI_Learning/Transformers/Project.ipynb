```python
#to check cuda and import pytorch (torch)
import torch
torch.cuda.is_available()
```

```python
from transformers import AutoModelForCausalLM
```

```python
model = AutoModelForCausalLM.from_pretrained(
    'google/gemma-2-2b-it',
    torch_dtype=torch.float16,
    device_map="auto",
    low_cpu_mem_usage=True, #to use cpu less, gpu more
    offload_buffers=True, #to offload the ,model buffers in gpu
    local_files_only=True #to use already downloaded local model
)
```

```python
tokens = tokenizer('Write python print syntax',return_tensors = 'pt').to('cuda')
```

```python
#this will generate different predictions
output = model(input_ids=tokens["input_ids"])
print(output)
```

```python
#to generate the real output token ids
gen_output=model.generate(tokens['input_ids'],attention_mask=tokens['attention_mask'],max_new_tokens=100)
print(gen_output)
```

```python
#to generate the token ids in real text format
tokenizer.batch_decode(gen_output)
```

```python
tokenizer.decode(gen_output)
```
