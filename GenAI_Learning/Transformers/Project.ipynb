```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
gemeni_embeding = GoogleGenerativeAIEmbeddings(model='gemini-embedding-2',output_dimensionality=150)
```

```python
query = "Hello, embedding the text"
vectordata = gemeni_embeding.embed_query(query)
print(f"Embedded data: {vectordata}")
print(f"Length for query vector data : {len(vectordata)}")
```
