#How LLMs Works

Large Language Models (LLMs) function as highly sophisticated predictive engines. They do not "think" or "understand" in the human sense; instead, they use advanced deep learning to analyze the statistical relationships between words (called tokens) and continuously predict what word should come next

### Transformer Architecture

------------------------------------- Encoder -------------------------------------

####Encoder - Its task is to collect the context of the input.

1-> Encoding :
    1. Tokenization : It tokenises the input data and associate integer value with the token. (tikTokenizer). It can be word by word or letter by letter tokenization. It depends on the model vocabulary size.
     If vocabulary size is huge -> word by word , and vice versa.

    2. Input Embedding : It is generated to capture semantic meaning of input.

2 -> Vector Embedding:
    1. To find the dimensional relation between the semantic meaning of input, if closer, more meaningfull, and vice versa.

3 -> Positional Encoding:
    - It is done so that we can maintain the order in the sentence, that how the word order is their in sentence. Which word comes after which word.
    - It adds more information to previos semantic meaning of word order.

4 -> Self Attention:
    - In case of Self attention layer, it tries to capture the context in which the word is used in the sentence, where and how the word is used, and what its context related to the sentence.
    - These embedings communicate with others to get to find out how the other emebedings are generated. (single head self attention)

5 -> Multi-Head Attention
    - It allows the model to look at the same sentence from multiple perspective at the same time.
    - Each attention head focuses on different types of relationship between the words.
    - These kind be different kind of heads:
        - What action affects
        - A long distance dependencies
        - Another is negation
    - By combining these views , the model forms a richer understanding of the sentence.
    - It is like reviwing the sentence by many experts like:
        - Grammer expert
        - Meaning expert
        - Relationship expert
        - Long context expert
        - Dependency of words expert
    - Each expert highlights diffrent connections between words

6 -> Normalization Layer , Feed Forward layer

------------------------------------ Decoder -------------------------------------

####Decoder:
    - Its task is to predict next word in the order sequentialy based on the input data passed to encoder.
    - There are two phase of decoder:
        - Training:
            - Here it predicts the next word using forward propogation.
        - Inferencing
    - Linear layer and softmax layer : Here different possiblities of prediction assigned with its probabilities number, and then it release the possibility with highest probability. We can even change
     the outcome using Temperature for creativity.

    -Input - Ex. Hello how are you?
        generation -  I (p1)(probability 0.5) or Ar(0.2) or B(0.4) or k(0.2)
        input to decoder - Hellow how are you? + p1 (highest probability posibility)
            generation - Hello how are yoy? I a(p2)
             Input to decoder - How are you? + p1 + p2
             .......these keeps on iteration till end of string meets.
