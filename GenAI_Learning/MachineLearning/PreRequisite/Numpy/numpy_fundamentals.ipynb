# Numpy

- Numpy is a fundamental package for scientific computing in python.
- It is a python library that provides multidimensional array objects and various derived objects such as Masked arrays or Matrices and a assortment of routines for fast operations on Arrays.
- It includes mathematical, Logical, Shape manipulation, sorting, selecting, I/O, Discrete Fourier Transform, Basic Linear Algebra, Stastical Operations, Random Simulations Etc.
- At the core of the Numpy Package is the dimensional Array object.

## Why Numpy ?

- Data types in python -> List and dictionary -> Slow
- We needed speed and less memory consumption to work on huge amount of data.
- Numpy was created in C language and wrapped in python.

- Libraries like Scikit Learn, Tensorflow, and Pytorch require input data in Numpy arrays or tensors

### Numpy Arrays Vs Python Sequences

- Numpy arrays have fixed size at creation, unlike python list, which grow dynamically.
- Changing the size of an ndim array will create a new array and delete the original
- The elements in a numpy array are required to be of the same data type and thus will be of the same size in memory
- Numpy arrays facilitate advanced mathematical and other types of operations on large number of data.
- Such Operations are executed more efficiantly with less code that is possible using python built in sequences.

### Functions in Numpy for Machine Learning and Data Science

- max() : It is used to calculate max value inside a matrix or an array
    -   Syntax
        - np.max(array)

- min() : It is used to calculate min value inside a matrix or an array
    -   Syntax
        - np.min(array)

- axis : It Provides us idea about any operation value inside specific row or column, Or to get any specific value from row wise or column wise.
    - axis=0 -> Applying operation column wise
    - axis=1 -> Applying operation row wise

- sum() : It calculates sum of all elements in matrix
    - Syntax
        - np.sum(array)

- prod() : It return the product of all elements in array.
    - Syntax
        - np.prod(array)

```python
arr = np.round(np.random.random((4,4))*50).astype("int32")
arr
```

```python
#to get the maximum value from the whole matrix
print(np.max(arr))
print(np.min(arr))
```

```python
#Column wise max value using axis=0
print(np.max(arr,axis=0))
#Row wise max value using axis=1
print(np.max(arr,axis=1))
```

```python
#Column wise min value using axis=0
print(np.min(arr,axis=0))
#Row wise min value using axis=1
print(np.min(arr,axis=1))
```

```python
# printing sum of all values in array
print(np.sum(arr))
```

```python
#Column wise sum value using axis=0
print(np.sum(arr,axis=0))
#Row wise sum value using axis=1
print(np.sum(arr,axis=1))
```

```python
# printing product of all values in array
print(np.prod(arr))
```

```python
#Column wise product value using axis=0
print(np.prod(arr,axis=0))
#Row wise product value using axis=1
print(np.prod(arr,axis=1))
```
