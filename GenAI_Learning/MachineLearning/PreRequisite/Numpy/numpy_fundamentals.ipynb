# Numpy

- Numpy is a fundamental package for scientific computing in python.
- It is a python library that provides multidimensional array objects and various derived objects such as Masked arrays or Matrices and a assortment of routines for fast operations on Arrays.
- It includes mathematical, Logical, Shape manipulation, sorting, selecting, I/O, Discrete Fourier Transform, Basic Linear Algebra, Stastical Operations, Random Simulations Etc.
- At the core of the Numpy Package is the dimensional Array object.

## Why Numpy ?

- Data types in python -> List and dictionary -> Slow
- We needed speed and less memory consumption to work on huge amount of data.
- Numpy was created in C language and wrapped in python.

- Libraries like Scikit Learn, Tensorflow, and Pytorch require input data in Numpy arrays or tensors

### Numpy Arrays Vs Python Sequences

- Numpy arrays have fixed size at creation, unlike python list, which grow dynamically.
- Changing the size of an ndim array will create a new array and delete the original
- The elements in a numpy array are required to be of the same data type and thus will be of the same size in memory
- Numpy arrays facilitate advanced mathematical and other types of operations on large number of data.
- Such Operations are executed more efficiantly with less code that is possible using python built in sequences.
