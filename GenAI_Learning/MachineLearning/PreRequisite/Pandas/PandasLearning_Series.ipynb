#### Creating Series in multiple ways
- List
- Numpy Array
- Dictionary
- Custom index
- Scalar Value

```python
# Series from list
country = ["India","USA","Germany","Italy"]
series2 = pd.Series(name="Country",data=country)
print(series2)
```

```python
itemName = ["Pen","Pencil","Notebook"]
itemPrice = [10,5,50]

series3 = pd.Series(name="Items Price",data=itemPrice,index=itemName,dtype="int32")
print(series3)
```

```python
# Series from NuMpy Array
values = np.arange(10,14)
series4 = pd.Series(name="Numpy array series",data=values)
print(series4)
```

- Dictionary to series
    - With dictionary we dont need to mention index expelicitly because dictionary itself comes with key value pair

```python
# Dictionary to series

dictvalue = {
    "Hindi" : 60,
    "English" : 70,
    "Maths" : 90
}

series5 = pd.Series(name="Subject marks",data=dictvalue)
print(series5)
```

```python
# Custom index in series
datalist = [40,45,78,41]

series6 = pd.Series(name="Custom series", data = datalist, index=["a","b","c","d"])
print(series6)
```

```python
# Scalar Value inside the series

series7 = pd.Series(name="Scalar value series",data=5,index=["a","b","c","d"])
print(series7)
```

## Important Series Methods

- head(n) --> It will return the first n data in the dataset. Defualt is 5 if no N provided.
- tail(n) --> It will return the last n data in the dataset. Defualt is 5 if no N provided.
- sample(n) --> It provide any custom row from the dataset. Default n will be 2
- value_counts() --> It provides the count of values (categorical values) getting repeated.
- sort_values() --> It sorts the values in the series in ascending manner default. for numerical columns
- sort_values(ascending=False) --> It sorts the values in the series in descending manner. for numerical columns
- copy() --> To copy the values of data to other variable
- sort_index() --> TO sort the data based on index sorting
