#### Map function map()

```python

lists = [1,2,3,4,5]
print(list(map(lambda x:x**2,lists)))
# print(nlist)
```

```python
li1 = [1,2,3]
li2 = [4,6,8]

print(list(map(lambda x,y:x+y,li1,li2)))
```

```python
# map() to convert all strings values items in an iterable into a integer

print(list(map(int,['1','2','5','7'])))
```

```python
#convert all names in uppercase

names = ["yash","rohan","vikas"]
print(list(map(str.upper,names)))
```

```python
#Retrieve values inside a dictionary using map()

employess = [{"name":"yash","age":26},
             {"name":"raj","age":29}
             ]

def agefinder(listitem):
    return listitem["age"]

print(list(map(agefinder,employess)))
```

### Filter function

```python
#retrieve only items which are even

def even(n):
    return n%2==0
```

```python
li = [1,2,3,4,5,6,7,8]
print(list(filter(even,li)))
```

```python
#retrieve only items which are even using lambda function
li = [1,2,3,4,5,6,7,8]
even_ = lambda x:x%2==0

print(list(filter(even_,li)))
```

```python
# retrieve the values greater that 5

li = [1,2,3,4,5,6,7,8]
greater_5 = lambda x:x>5

print(list(filter(greater_5,li)))
```

```python
# retrieve the values greater that 5 and even

li = [1,2,3,4,5,6,7,8]
greater_even = lambda x:x>5 and x%2==0
print(list(filter(greater_even,li)))
```

```python
#Retrieve values inside a dictionary using filter()

employess = [{"name":"yash","age":24},
             {"name":"raj","age":29}
             ]

def agecheck(item):
    return item["age"] > 25

print(list(filter(agecheck,employess)))
```

```python
# reading total lines, total words. total characters from a file ->> readlines()

fp = r"C:\Users\yashr\PycharmProjects\Python-Automation-Code-Practice\assets\test"

def count_lines(fp):
    with open(fp,"r") as f:
        lines = f.readlines()

        print(f"Lines count : {len(lines)}")
        totalwords = sum(len(line.split()) for line in lines)
        print(f"Word count : {totalwords}")
        totalchars = sum(len(line) for line in lines)
        print(f"Total Characters : {totalchars}")

count_lines(fp)
```
