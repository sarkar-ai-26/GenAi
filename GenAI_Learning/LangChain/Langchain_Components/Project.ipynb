```python
#to load the env parameters from .env file, secret keys
from dotenv import load_dotenv
#Loading the .env file secret keys
load_dotenv()
```

```python
from langchain_huggingface import HuggingFacePipeline
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch
```

```python
model_id = 'google/gemma-2-2b-it'
```

```python
#creating object that can return the tokens of the inputs
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    device_map="auto",
    low_cpu_mem_usage=True, #to use cpu less, gpu more
    local_files_only=True #to use already downloaded local model
)
```

```python
# transformers pipeline
pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=256,
    temperature=0.7
)
clean_up_tokenization_spaces=False
# Wrap it in LangChain
local_llm = HuggingFacePipeline(pipeline=pipe)

response = local_llm.invoke("Write a quick python script to automate file backups.")
print(response)
```
