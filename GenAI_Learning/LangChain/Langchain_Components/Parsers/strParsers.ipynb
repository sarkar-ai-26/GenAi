```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()
```

```python
gemeni_model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite",temperature=0.2)
llama_model = ChatOllama(model='llama3.1:8b',temperature=0.5)
```

```python
prompt_1 = PromptTemplate(template='Create a detailed report for this {Topic}',
                          input_variables=['Topic'])
```

```python
prompt_2 = PromptTemplate(template='Generate a 5 point summary on this {Report}',
                          input_variables=['Report'])
```

```python
strParser = StrOutputParser()
```

```python
chain = prompt_1 | gemeni_model | strParser | prompt_2 | llama_model | strParser
```

```python
final_responce = chain.invoke({'Topic':'Electric Cars Vs Combustion Engine Cars'})
```
