```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser

from GenAI_Project.ChatBots.DynamicChatBot.RGC_ChatBot import template
```

```python
load_dotenv()
```

```python
g_model = ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite',temperature=0.5)
l_model = ChatOllama(model='llama3.1:8b',temperature=0.5)
```

```python
g_template = PromptTemplate(template='''Create a 100 word Analysis summary on this topic {Topic}''',input_variables=['Topic'])
```

```python
l_template = PromptTemplate(template='Create a 5 point summary from this analysis {Analysis}',input_variables=['Analysis'])
```

```python
parser = StrOutputParser()
```

```python
#This is Langchain Expression Language
chain = g_template | g_model | parser | l_template | l_model | parser
```

```python
responce = chain.invoke({"Topic":'Electric vehicle'})
```
