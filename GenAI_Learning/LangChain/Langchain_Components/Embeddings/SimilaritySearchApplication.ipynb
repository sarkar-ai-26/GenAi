```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
embedding = GoogleGenerativeAIEmbeddings(model='gemini-embedding-2',output_dimensionality=200)
```

```python
docsList = ['Hello this is first GenAI embedding sentence','Here i am learning the concept of embedding','And i am also learning the practical usage with gemeni-embedding-2 model']
query='What user is learning here?'
```

```python
docsListEmb =embedding.embed_documents(docsList)
queryEmb= embedding.embed_query(query)
```

```python
score = cosine_similarity([queryEmb],docsListEmb)[0]
```

```python
index, value =sorted(list(enumerate(score)),key=lambda x:x[1])[-1]
```

```python
print(query)
answer = docsList[index]
print(answer)
print(f"Similarity score is : {value}")
```
