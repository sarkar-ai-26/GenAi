```python
query="How GenAI helps in industry?"
```

```python
vec_store = FAISS.from_documents(dataset,embed)
```

```python
multiretriever=MultiQueryRetriever.from_llm(retriever=vec_store.as_retriever(search_type='mmr',search_kwrgs={"k":2,'lambda_mult':0.5}),llm=LLM)
```

```python
results = multiretriever.invoke(query)
```
