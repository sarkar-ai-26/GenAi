### Problem with LLM to use directly?
- LLM model don't remember context beyond the prompt.
- It doesn't have the context for the next follow-up queries.
- They can not use directly external data sources like, apis, databases, etc.
- LLM models cant perform actions like searching the web, running code., or queries SQL by themselves.

### What is LangChain?

- It's an opensource framework that helps in building LLM based application.
- It provides modular components or end to tools like data collecting data, data injetion, connecting data and more, that helps developer build complex AI application. Such as ChatBots, Workflow Agents, Query bots, RAG application, Autonomous AI agents, AI knowledge assistant.

### Features of Langchain

- It supports all major LLMs like OpenAi GPT, Google Gemeni, Gemma, Hugging face models, LLAMA, Mistral etc.
- This framework helps us in connecting all these LLMs with our own data.
- It simplifies development LLM based application as it is having major tools as
    - chains : it chains together all tools, and that why complex application requireing multiple app together can be easily build.
        
- It is opensource.
- It can manage complex workflows, where llms interact with databases, 3rd party softwares, mcp servers and more.
- Its a bridge between LLMs and real world data and application, and to chain all together.

## Prompting Frameworks

> **Standard Structure : Role -> Goal -> Context -> Example -> Output Structure**

- Shot prompting : In this we need to provide example to model to how to generate response data.
    - Zero shot prompting : No previous data provide, just question.
    - One shot prompting : In this we need to give one example for response.
    - Few shot prompting : In this we need to provide more than 1 examples to create the context of the prompt.
- Chain of thoughts : This helps models to explain its answers in a step by step format. Ex, Lets think step by step, behind the solution.
- Ask before answer technique : Model will ask user for the options or method to go forward with the response.
- Fill in the blanks prompting : User will ask model to provide bullet points questions? and then user will provide the answer with the points mentioned, and then model will do the next response generation.
- Reverse Prompting : Reverse engineering, such that user will have the response or content already, and then user can ask query to generate the prompt used.
- RGC Prompting : **Role -> Goal -> Context** -> Result -> Constraint

# Problem with load -> it will load all files content in the memory, and it will take lot of time


> load() -> It is eager loading, as it loades everything at once and return a list of document object, That means, if we have very less document then it is okay, but when the data is huge, files are huge, then its very time and memory consuming module.

> lazy_load() -> It load with demand basis. And return a generator document object. So all documents are not loaded and once, it loads as per requirement. It is used for large documents, or lot of files.

> In LangChain, lazy loading (.lazy_load()) is a memory-efficient technique used in Document Loaders to process and stream large datasets. .

> Instead of loading all documents into memory at once as a list (which causes out-of-memory errors), .lazy_load() yields documents one by one using a Python generator.

> Why use .lazy_load()?
- Low Memory Footprint: Ideal for handling thousands of PDFs, large CSVs, or massive database dumps.
- Streaming & Chinking: You can chunk, embed, or send documents to a vector store as soon as a single document is yielded, releasing memory before the next one is processed.How to use itMost modern LangChain document loaders support both .load() (eager) and .lazy_load()

## Text - Splitters

- It devides large documents into smaller chunks of text.
- These chunk can be used for embedding, semantic search, and for RAG.
- It also helps in downstream tasks
    - Embedding -> short chunks yield more accurate vectors.
    - Semantic search
- To get rid of hallucination.
- To optimize computation resouces, and make more memory efficient, and allow more parallel processing.
- Types:
    - Length Based
    - Text Structure Based
    - Document Structure Based
    - Semantic Meaning based



## Length Based Text Splitter

- It is the fastest text splitting technique and easiest.
- Here it splits the data where we already decide that how many characters or tokens we need to have in every chunks.
- **Not Recomeneded** (chunks can be incomplete in the starts and ends of chunks).
- It can not capture linguistic meaning and semantic meaning.

> from langchain_text_splitters import CharacterTextSplitter

> textSplit = CharacterTextSplitter(chunk_size=100,chunk_overlap=20,separator='')

> chunks = textSplit.split_text(''' '''')

>chunks[0]

- Parameters:
- chunk_size : chunk size in each chunk
- chunk_overlap : Overlap b/w chunks to maintains the chunks. It can be 10% of chunk size.
- separators : Any Separators between the text data.

### Retriever

- Vector Store <- Fetching the relevant content for us for answering users query.
- LLM would only rely on its training data.
- LLM doesnt have access to private data.
- With retriever, model gets most relevant and fresh private data.
- A retriever in langchain component fetches the most relevant information (chunks) from the knowledge base that is vector store/database for a users query.
- Retriever alone can not provide answer in proper way. To generate a answer in proper way, LLM is used.

>Workflow of Retrieval

-> User query -> Retriever converts query into embeddings -> Then Retriever searches vector store for similar chunks -> It return the top most relevant content of the documents. -> It passed to LLM with query (Augmentation) -> LLM generate the proper data.



> Types of Retriever:
1. DataSource (Vector store)
2. APIs, Ex (WikiPedia)
3. Search-Strategy -> VectorStore Search Vs Retriever Search
4. Maximum Marginal retriever
5. Multi Query retriever
6. Contextual Compression Retriever

### Wikipedia Retriever

### Vector Store Retriever

- DataSource -> Vector Store -> Retriever -> LLM
- Langchain vector store retriever is the most common type of retriever that search and fetch the most relevant documents from the vector store based on semantic similarity using vector embedding.


### Maximal Marginal Relevance Retriever

- It is an information retrieval algorithm designed to reduce the redundancy in the Retrieved data while maintaining high relevance to the query.
- It is to pick result that are not only relevant to the query but also different from each other.
- In regular similarity search documents can be very similar to each other, lacking diversity in perspective.
- MMR retrieve avoids by picking the most relevant doc first.
- Then it picks the next most relevant but least similar to priveos one.
