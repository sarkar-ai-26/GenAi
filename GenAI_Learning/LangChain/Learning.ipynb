# LangChain

> Roles That comes under : (AI Developer, LLM Engineer. GenAi Engineer, RAG Developer)

- We can create LLM applications using already available LLMs.

### Problem with LLM to use directly?
- LLM model don't remember context beyond the prompt.
- It doesn't have the context for the next follow-up queries.
- They can not use directly external data sources like, apis, databases, etc.
- LLM models cant perform actions like searching the web, running code., or queries SQL by themselves.

### What is LangChain?

- It's an opensource framework that helps in building LLM based application.
- It provides modular components or end to tools like data collecting data, data injetion, connecting data and more, that helps developer build complex AI application. Such as ChatBots, Workflow Agents, Query bots, RAG application, Autonomous AI agents, AI knowledge assistant.

### Features of Langchain

- It supports all major LLMs like OpenAi GPT, Google Gemeni, Gemma, Hugging face models, LLAMA, Mistral etc.
- This framework helps us in connecting all these LLMs with our own data.
- It simplifies development LLM based application as it is having major tools as
    - chains : it chains together all tools, and that why complex application requireing multiple app together can be easily build.
        
- It is opensource.
- It can manage complex workflows, where llms interact with databases, 3rd party softwares, mcp servers and more.
- Its a bridge between LLMs and real world data and application, and to chain all together.

### Why LangChain and benefits?

- Memory : That keeps track of the conversation and context.
- Data-connector : It helps is connecting any kind of database like, AWS, GCP, Azure, SQL, APIs, Google Drive, Localdrive.
- Data Loaders : It helps in loading different kind of data.
- Tools & Agents : Let Llms takes action on their own., Browsing capability, python code execution, check 3rd party softwares and grab functionalities.
- Chains : It helps in structuring these workflows.
- Model Agnostic Development -> We can shift from one model to another, or databases, or apis.
- It Provides complete Ai Ecosystem.
- Langsmith : To deploy the application or convert the application in API so it can be used from out source.

## Prompting Frameworks

> **Standard Structure : Role -> Goal -> Context -> Example -> Output Structure**

- Shot prompting : In this we need to provide example to model to how to generate response data.
    - Zero shot prompting : No previous data provide, just question.
    - One shot prompting : In this we need to give one example for response.
    - Few shot prompting : In this we need to provide more than 1 examples to create the context of the prompt.
- Chain of thoughts : This helps models to explain its answers in a step by step format. Ex, Lets think step by step, behind the solution.
- Ask before answer technique : Model will ask user for the options or method to go forward with the response.
- Fill in the blanks prompting : User will ask model to provide bullet points questions? and then user will provide the answer with the points mentioned, and then model will do the next response generation.
- Reverse Prompting : Reverse engineering, such that user will have the response or content already, and then user can ask query to generate the prompt used.
- RGC Prompting : **Role -> Goal -> Context** -> Result -> Constraint
