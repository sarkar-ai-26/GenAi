```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
cricketer_dataset = [
    Document(
        page_content="Sachin Tendulkar is widely regarded as one of the greatest batsmen in the history of cricket. He is the all-time highest run-scorer in both Test and ODI cricket, and the only player to score one hundred international centuries.",
        metadata={"player_name": "Sachin Tendulkar", "country": "India", "role": "Batsman"}
    ),
    Document(
        page_content="Sir Don Bradman was an Australian international cricketer, widely acknowledged as the greatest batsman of all time. His career Test batting average of 99.94 is claimed to be the greatest achievement by any sportsman in any major sport.",
        metadata={"player_name": "Don Bradman", "country": "Australia", "role": "Batsman"}
    ),
    Document(
        page_content="Virat Kohli is a contemporary right-handed batsman known for his aggressive style and exceptional chasing ability in limited-overs cricket. He holds the record for the most centuries in One Day Internationals.",
        metadata={"player_name": "Virat Kohli", "country": "India", "role": "Batsman"}
    ),
    Document(
        page_content="Wasim Akram is a former Pakistani cricketer and captain, widely considered one of the greatest fast bowlers in cricket history. He was a master of swing bowling and the first bowler to reach 500 ODI wickets.",
        metadata={"player_name": "Wasim Akram", "country": "Pakistan", "role": "Bowler"}
    ),
    Document(
        page_content="Sir Garfield Sobers is universally recognized as cricket's greatest ever all-rounder. He was a highly skilled batsman, an effective bowler utilizing multiple styles, and an exceptional fielder.",
        metadata={"player_name": "Garfield Sobers", "country": "West Indies", "role": "All-rounder"}
    )
]
```

```python
cricketer_dataset
```

```python
vector_store = Chroma(embedding_function=OllamaEmbeddings(model="embeddinggemma"),persist_directory='chroma_database',collection_name='smpledata')
```

```python
vector_store.add_documents(cricketer_dataset)
```

```python
vector_store.get(include=['documents','embeddings','metadatas'])
```

```python
vector_store.similarity_search(query='Who is the greatest all rounder in cricket history ?', k=1)
```

```python
vector_store.similarity_search_with_score(query='Who is the greatest all rounder in cricket history ?', k=1)
```
