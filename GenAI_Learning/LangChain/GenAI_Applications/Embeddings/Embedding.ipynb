```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
google_embedding = GoogleGenerativeAIEmbeddings(model='gemini-embedding-2',output_dimensionality=250)
```

```python
result = google_embedding.embed_query("Hi this is my first embedding text") #embed_query for single statement
print(result)
```

```python
#For collection of statements
docsList = ['Hello this is first GenAI embedding sentence','Here i am learnig the concept of embedding','And i am also learning the practical usage with gemeni-embedding-2 model']
```

```python
resultDocs = google_embedding.embed_documents(docsList)
print(resultDocs)
```
