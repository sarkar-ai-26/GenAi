```python
from dateutil.rrule import weekday


#class object syntax
class Avengers:
    # __jnit__ is a costructor initializer function
    def __init__(self,name,power,weapon,citizen="American"): #attributes
        self.name = name
        self.power = power
        self.weapon = weapon
        self.citizen = citizen

    def introduction(self):
        print(f"Hello people, my name is {self.name}")
        print(f"My power is {self.power}")
        print(f"My weapon is {self.carrying_weapon()}")
        print(f"My citizen is {self.citizen}")

    def carrying_weapon(self):
        return self.weapon
```

### Cartoon Characters

```python
class Cartoon:
    def __init__(self,name, show, catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase

    def intro(self): #instance method
        print(f"Name : {self.name}")
        print(f"Show : {self.show}")
        print(f"My best catchphras : {self.speak()}")

    def speak(self):
        return self.catchphrase
```

### Types of Methods:
- Instance Method -> instance methods operates on objedct attributes, it's going to use self keyword to access and modify objects attributes.

- Class Method -> It is the method of the whole class or the whole blueprint which will remian the same for all the objects then such kind of method is being reffered as a Class methods.
        - This method will work with the class level. They are going to use <cls> instead of self.
        - They are called using the class name as well.
        - Need to use @classmethod

- Static Method -> It will not use self, or cls,. It will work as normal method, under class. Decorator <@staticmethod>


### Class Method

```python
class MainClass:

    origin = "Version200"

    def __init__(self,first,second):
        self.first = first
        self.second = second
    def intro(self):
        print(f"First : {self.first}, Second : {self.second}")

    @classmethod
    def statingorigin(cls):
        print(f"Origin : {cls.origin}")
```

```python
firstobjest = MainClass("Hello","Boy")
firstobjest.intro()
```

```python
firstobjest.statingorigin()
```

### Inheritance


## Single Inheritance

```python
class Avenger:
    def __init__(self,name,power):
        self.name = name
        self.power = power

    def intro(self):
        print(f"Name :{self.name}")
```

```python
class Xmen(Avenger): #need to mention the parent class name, it can have multiple class inside it
    def __init__(self,name,power,weapon):
        super().__init__(name,power)
        self.weapon = weapon

    def weapondetails(self):
        print(f"Weapon : {self.weapon}")
```

```python
iron = Xmen("IronMan","IronSuit","Armour")
```

```python
iron.name
```

```python
iron.weapondetails()
```

```python
iron.power
```

## MultiLevel Inheritance

```python
class GrandParent:
    def __init__(self,race,nationality,skin):
        self.race = race
        self.nationality = nationality
        self.skin = skin

    def intro(self):
        print(f"Race :{self.race}, Nationality : {self.nationality}, Skin : {self.skin}")
```

```python
class Parent(GrandParent): #need to mention the parent class name, it can have multiple class inside it
    def __init__(self,race,nationality,skin,wealth):
        super().__init__(race,nationality,skin)
        self.wealth = wealth
    def wealthdetails(self):
        print(f"Wealth : {self.wealth}")
```

```python
class Child(Parent): #need to mention the parent class name, it                  can have multiple class inside it
    def __init__(self,race,nationality,skin,wealth, study):
        super().__init__(race,nationality,skin,wealth)
        self.study = study
    def studydetails(self):
        print(f"Study : {self.study}")
```

```python
study = Child("Asian","Indian","Brown","Rich","Btech")
```

```python
# print(study.race)
# print(study.nationality)
print(study.wealthdetails())
```

```python
print(study.intro())
```

```python
study.nationality = "German"
```

```python
study.intro()
```

## Magic methods

They have double underscore before and after the name. They also knowns as dunder method.
They are special methods are in python that allows us to customize how objects behave
Makes objects behave like built in types (eg,list,integers)
Improves Readability by allowing basic operations


They can help in:
- Object Intialization (__init__)
- String Representation (__str__,__repr__)
- Arithmetic Operations (__add__,__sub__ etc)
- Comparison Operation (__eq__)
- Length,indexing, iteration (__len__,__getitem__)
