## Task 1 : basic Functions: Price After Discount


```python
from urllib.parse import uses_relative


def apply_discount(price,discount_percent = None):
    if discount_percent == None:
        discount_percent = 5
    elif discount_percent > 60:
        print("Discount can not exceeds 60%")
        return

    priceAfterDiscount = price - (price * (discount_percent/100))

    return priceAfterDiscount

price = apply_discount(1000,10)
print(price)
```

## Task 2 - Resursive Function : Factorial utility


```python
def fact(n):
    if n < 0:
        print("N is negative")
        return None
    elif n == 0 or n == 1:
        return 1

    return n * fact(n-1)


n = -8
facto = fact(n)
print(f"Factorial of {n} is {facto}")
```

## Task 3 - Lambda Function: GST Calculator



```python
gst = lambda price:price + (0.18 * price)
print(gst(100))
finalprice = lambda price,discount:gst(price) - (gst(price) * (discount/100))

print(finalprice(100,10))
```

## Task - 4 - Using map(): Applu GST to list of Prices


```python
prices = [100,250,400,1200,50]
print(f"Original Price : {prices}")
price_with_gst = list(map(gst,prices))
print(f"Price with GST: {price_with_gst}")
```

## Task - 5 - Using filter() : Filter Expensive Products


```python
prices = [100,250,400,1200,50,2000,850]
greaterThan500 = lambda price : price > 500
lessThanEqual500 = lambda price: price <= 500

print(list(filter(greaterThan500,prices)))
print(list(filter(lessThanEqual500,prices)))
```

## Task 6 - Combined utility Function


```python
def process_prices(prices):
    dis = lambda price:price - (price * (10/100))
    discount_price = list(map(dis,prices))
    disfilter = lambda price:price>500
    filterPrice = list(filter(disfilter,discount_price))
    return discount_price, filterPrice

prices = [100,500,900,50,750]
discount_price, filterPrice = process_prices(prices)
print(f"Discounted price list : {discount_price}\n"
      f"Filtered Price List : {filterPrice}")
```

## Task 7 - Mini Problem: Menu using Function


```python
def add_price(prices_list,price):
    return prices_list.append(price)

def get_average_price(prices_list):
    return sum(prices_list) / len(prices_list)

def get_max_price(prices_list):
    return max(prices_list)

price_list = []
while True:
    user_input = input(f"1. Add price \n"
          f"2. Show Average price \n"
          f"3. Show Highest price \n"
          f"q. press q to QUIT")
    if user_input == "1":
        price = int(input("Enter the new price : "))
        add_price(price_list,price)
    elif user_input == "2":
        avg = get_average_price(price_list)
        print(f"Average Price : {avg}")
    elif user_input == "3":
        maxx = get_max_price(price_list)
        print(f"Maximum Value : {maxx}")
    elif user_input == "q":
        break
    else:
        continue
```
