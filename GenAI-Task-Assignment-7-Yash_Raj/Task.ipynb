### Task 1 : Basic Class & Object Creation


```python
from tkinter import PhotoImage


# creating class with name Product
class Product:
    def __init__(self,name,price,category,discount):
        self.name = name
        self.price = price
        self.category = category
        self.discount = discount

    def get_info(self):
        print(f"Product Details : \n"
              f"Name - {self.name}\n"
              f"Price - Rs. {self.price}\n"
              f"Discounted Price - Rs. {self.apply_discount()}\n"
              f"Category - {self.category}")
    #optional task to apply discount , this will help in returning discounted price
    def apply_discount(self):
        return self.price - (self.price * self.discount/100)
```

```python
product_1 = Product("Pen",10,"Study",10)
product_2 = Product("Phone",10000,"Electronics",20)
```

```python
print(product_1.get_info())
print(product_2.get_info())
```

### Task 2 : Constructor and Encapsulation


```python
class Product:
    def __init__(self,price):
        self.__price = price

    def get_price(self):
        print(f"Price - Rs. {self.__price}")

    def set_price(self,new_price):
        if new_price > 0:
            self.__price = new_price
        return self.get_price()
```

```python
product_1 = Product(20)
```

```python
product_1.get_price()
```

```python
product_1.set_price(30)
```

### Task 3 : Inheritance (Single-Level)


```python
class Product:
    def __init__(self,name,price,category,discount):
        self.name = name
        self.price = price
        self.category = category
        self.discount = discount

    def get_info(self):
        print(f"Product Details : \n"
              f"Name - {self.name}\n"
              f"Price - Rs. {self.price}\n"
              f"Discounted Price - Rs. {self.apply_discount()}\n"
              f"Category - {self.category}")

    def apply_discount(self):
        return self.price - (self.price * self.discount/100)

```

```python
class ElectronicProduct(Product):
    def __init__(self,name,price,category,discount,warranty_years):
        super().__init__(name,price,category,discount)
        self.warranty_years = warranty_years

    def get_info(self):
        print(f"Product Details : \n"
              f"Name - {self.name}\n"
              f"Price - Rs. {self.price}\n"
              f"Discounted Price - Rs. {self.apply_discount()}\n"
              f"Category - {self.category}"
              f"Warranty Years - {self.warranty_years}")
```

```python
elecPro = ElectronicProduct("Phone",1524,"Electronic",25,5)
elecPro.get_info()
```

### Task 4 - Polymorphism


```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price

    def get_info(self):
        print(f"Product Details : \n"
              f"Name - {self.name}\n"
              f"Price - Rs. {self.price}\n")
```

```python
class Laptop(Product):
    def __init__(self,name,price,ram):
        super().__init__(name,price)
        self.ram = ram

    def get_info(self):
        print(f"Product Details : \n"
              f"Name - {self.name}\n"
              f"Price - Rs. {self.price}\n"
              f"Ram - {self.ram}")

class Mobile(Product):
    def __init__(self,name,price,camera):
        super().__init__(name,price)
        self.camera = camera

    def get_info(self):
        print(f"Product Details : \n"
              f"Name - {self.name}\n"
              f"Price - Rs. {self.price}\n"
              f"Camera - {self.camera}")
```

```python
laptop = Laptop("Asus",20000,"12GB")
phone = Mobile("Samsung",52000,"50MP")
```

```python
laptop.get_info()
phone.get_info()
```

### Task 5 - Abstarction (Using Abstarct Base class)


```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass
```

```python
class CreditCardPayment(Payment):
    def process_payment(self,amount):
        print("Amount ", amount + 10 )
```

```python
class UPIPayment(Payment):
    def process_payment(self,amount):
        print("Amount ", amount + 100 )
```

```python
cp = CreditCardPayment()
cp.process_payment(20)
```

```python
upip = UPIPayment()
upip.process_payment(50)
```

### Magic Methods & Operator Overloading


```python
class Products:
    def __init__(self,name,price):
        self.name = name
        self.price = price

    def __add__(self,other):
        return Products(self.name + "&" + other.name, self.price + other.price)

    def __str__(self):
        return f"Product name : {self.name} , Price of both : {self.price}"

```

```python
p1 = Products("Pen",100)
p2 = Products("Pencil",120)

both = p1 + p2
print(both)
```

### Task 7 - Mini Project: Simple Inventory System



```python
from types import SimpleNamespace

class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self,product):
        return self.products.append(product)

    def remove_product(self,name):
        for product in self.products:
            if product.name == name:
                self.products.remove(product)

    def get_total_value(self):
        return sum(p.price for p in self.products)

    def show_all_products(self):
        for item in self.products:
            print(f"Product : {item.name} , Price : Rs {item.price}")
```

```python
class Store(Inventory):
    def __init__(self,store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    def add_new_product(self):
        name = input("Enter new product name : ")
        price = int(input("Enter product price : "))
        product = SimpleNamespace(name =name,price=price)
        self.inventory.add_product(product)

    def show_summary(self):
        print(f"Total Items : {len(self.inventory.products)}")
        print(f"Total products value : {self.inventory.get_total_value()}")

store = Store("Yash-Store")
for _ in range(3):
    store.add_new_product()

store.show_summary()

```
