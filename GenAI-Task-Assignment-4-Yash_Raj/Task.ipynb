### Assignement 4 : File Handling


### Task 1 : Write Sales Records to a File


```python
#Creating list of sales
sales = [1200,450,980,1500,3000]
#opening the text file to add the data from the sales list
with open("sales_data.ext","w") as f:
    for item in sales:
        f.write(f"{(item)}\n")

#reading the file content after opening the file
with open("sales_data.ext","r") as file:
    data = file.read()

print(f"Data is file : {data}")

# opening the file and storing the data of sales list in file is comma-separated
with open("sales_data.ext","w+") as fcomma:
    for item in sales:
        fcomma.write(f"{item} ,")
    fcomma.seek(0)
    datacomma = fcomma.read()

print(datacomma)
```

### Task 2 : Read file in deiffrent ways


```python
#opening the file and reading the entire content
with open("sales_data.ext","r") as f:
    entire_data =f.read()
    #reading the first line of the file
    f.seek(0)
    first_line = f.readline()
print(f"Entire data in file : {entire_data}")
print(f"Entire first line in file : {first_line}")

#read all line and convert them into list of integers
with open("sales_data.ext",'r') as lines:
    alllines = lines.readlines()

print(f"All lines of data : {alllines}")
intlist=[]

for item in alllines:
    # new = new + item
    items = item.strip().split(',')
    items.pop(-1)

    for item in items:
        intlist.append(int(item))

print(f"Items in list as string : {items}")
print(f"Items in list as int : {intlist}")
```

### Task 3 : Append New sales


```python
# appending new sales values to the file
newdata = [5000,2500,1700]
with open("sales_data.ext","a+") as ap:
    for item in newdata:
        ap.write(f"{item}\n")
    ap.seek(0)
    data = ap.read()
    ap.seek(0)
    countline = ap.readlines()
print(f"All data : {data}")
print(f"Total Lines : {len(countline)}")
```

### Task 4 : Genrate Summary Report from file


```python
#read all line and convert them into list of integers
with open("sales_data.ext",'r') as lines:
    alllines = lines.readlines()

intlist=[]

for item in alllines:
    items = item.strip().split()

    for item in items:
        intlist.append(int(item))

print(f"Items in list as int : {intlist}")
#Total Sales
print(f"Total Sales : {sum(intlist)}")
#Highest Sale
print(f"Highest Sale : {max(intlist)}")
#Lowest Sale
print(f"Lowest Sale : {min(intlist)}")
#Average Sale
print(f"Average Sale : {sum(intlist)/len(intlist)}")
```

### Task 5 : Create product info file (User Input)


```python
# Asking user to provide Product and prices, Limit to 3
n = 3
productDict = {}
while n > 0:
    product = input("Enter Product Name : ")
    productPrice = int(input("Enter Price : "))
    productDict[product] = productPrice
    n = n - 1

print(f"Product Dict : {productDict}")

#Creating product.ext to add the product and their prices provided by user
with open("product.txt","a+") as file:
    for product, price in productDict.items():
        file.write(f"{product} | {price}\n")
        file.seek(0)
        data = file.read()

print(f"Data in product.txt : {data}")
```

### Task 6 : Read File Safely ( Error Handling inside File Handling Only )



```python
import os
#Taking user input for file name
filename = input("Enter file name ,Format 'filename.txt' : ")

if os.path.exists(filename):
    print(f"File Exists")
    with open(filename,'r') as f:
        data = f.read()
    print(f"Data in file : {data}")
else:
    print("File does not exist, check file name!")
```

### Task 7 : Mini Project - Export Discouted Prices


```python
prices = {
    "Mouse":500,
    "keyboard":800,
    "Monitor":7000,
    "Pendrive":400,
    "Camera":5000
}
#Asking user to provided discount
discount = int(input("Enter discount Percentage Ex. '5' : "))
discountprice = []
with open("discount_report.txt","w+") as file:
    for product, price in prices.items():
        #add product price and discounted value in required format X|X|X
        file.write(f"{product} | {price} | {price - (price * (discount/100))}\n")
        file.seek(0)
        #reading file
        data = file.read()
        #appending price in discount list to find average
        discountprice.append(price)

print(f"Data in product.txt : \n{data}")
print(f"Total items : {len(prices)}")
print(f"Average Discounted Price : {sum(discountprice)/len(discountprice)}")
```
