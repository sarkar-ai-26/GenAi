### Task 1 : Creating NumPy Arrays.


```python
import numpy as np
```

### Task 2 : Important Mathematical Operations



```python
A = np.array([10,20,30,40])
B = np.array([1,2,3,4])

# Performing Addition (A+B)
# add  = np.add(A,B)
add  = A+B
print(f"Addition of array A&B: {add}")

# Performing Substraction (A-B)
# sub  = np.subtract(A,B)
sub  = A-B
print(f"Subtraction of array A&B: {sub}")

# Performing Multiplication (A-B)
# mul  = np.multiply(A,B)
mul  = A*B
print(f"Multiplication of array A&B: {mul}")

# Performing Division (A/B)
# div  = np.divide(A,B)
div  = A/B
print(f"Divison of array A&B: {div}")

# Performing Power (A**B)
# powe  = np.power(A,2)
powe  = A**2
print(f"Divison of array A&B: {powe}")

```

### Task 3 : Important Numpy Mathematical Formulas


```python
values = np.array([2,4,6,8,10])

# Square root of each element
print(np.sqrt(values))

# Exponential root of each element
print(np.exp(values))

# Natural Log root of each element
print(np.log(values))

# Sum of all element
print(np.sum(values))

# Commulative Sum of all element
print(np.cumulative_sum(values))
```

### Task 4 : Aggregation Operation



```python
data = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

# Row-wise sum
print(np.sum(data,axis=1))

# Column-wise sum
print(np.sum(data,axis=0))

# Minimum Value
print(np.min(data))

# Maximum Value
print(np.max(data))

# Mean Value
print(np.mean(data))
```

### Task 5 : Statistical Operations


```python
marks = np.array([78,85,90,66,72,88,95,60])

#Mean
print(np.mean(marks))

#Median
print(np.median(marks))

#Varience
print(np.var(marks))

#Standard Deviation
print(np.std(marks))

#Min-Max Deviation
print(f"Min : {np.min(marks)}, Max : {np.max(marks)}")

# Range (max-min)
print(f"Range : {np.max(marks) - np.min(marks)}")

```

### Task 6 : Percentile & Sorting




```python
marks = np.array([78,85,90,66,72,88,95,60])

# Sort the array
sortedmarks = np.sort(marks)
print(f"Sorted Marks :  {sortedmarks}")

#25th percentile
# print(f"25th Percetile : {sortedmarks[round((25/100)*(len(sortedmarks)-1))]}")
print(f"25th Percetile : {np.percentile(marks,25,method="nearest")}")

#50th percentile
print(f"50th Percetile : {np.percentile(marks,50,method="nearest")}")

#75th percentile
print(f"75th Percetile : {np.percentile(marks,75,method="nearest")}")

#Count Students scored above average marks
avg = np.average(marks)
count = 0
for item in marks:
    if item > avg:
        count += 1
print(f"Count of students scored above average : {count}")
```

### Task 7 : Mini Use Case: Sales Analysis


```python
sales = np.array([1200,1500,900,2000,1800,1700,1600])

#Total weekly sales
print(f"Total Weekly sales : Rs. {np.sum(sales)}")

#Average daily  sales
print(f"Average daily sales : Rs. {np.average(sales)}")

#Highest and lowest sales day
print(f"Highest sales day : Rs. {np.max(sales)}, Lowest sales day : Rs. {np.min(sales)}")

#Standard deviation of sales
print(f"Standard deviation of sales : {np.std(sales)}")

#Days where sales above average
avg = np.average(sales)
count = 0
for item in sales:
    if item > avg:
        count += 1
print(f"ays where sales above average : {count}")
```
