### Task 1 : Safe Division Utility


```python
numerator = input("Enter the numerator")
denominator = input("Enter the denominator")
```

```python
try:
    devision = int(numerator)/int(denominator)
except ZeroDivisionError as zd:
    print(zd)
except ValueError as ve:
    print(ve)
else:
    print(devision)
finally:
    print("Operation complete")
```

### Task 2: Bill Calculator with Error Handling


```python
prices = [120,350,'abc',500,-200,800]
total = 0
for item in prices:
    try:
        if isinstance(item, str):
            raise TypeError("Invalid Value")
        elif item < 0:
            raise ValueError("Negative number not allowed")
        else:
            total+=item
    except TypeError as te:
        print("TypeError :",te)

    except ValueError as ve:
        print("ValueError :",ve)

print(f"Total price : {total}")
```

### Task 3 : Custom Exception : Age Validator


```python
#method to check age
def check_age(age):
    try:
        if not 1 <= age <= 120:
            raise ValueError("Age must be between 1 and 120")
    except ValueError as ve:
        print(f"ValueError : {ve}")
    else:
        return age
```

```python
#treatint block as my main code to call the check_age() method
age = int(input("Enter your age"))
check_age(age)
```

### Task - 4 - File Reader with Exception handling


```python
filename = input("Enter file name, ex file.text : ")
```

```python
import os
try:
    if not os.path.exists(filename):
        raise FileNotFoundError("File Not available")
    else:
        try:
            with open(filename,"r") as f:
                data = f.read()
        except PermissionError as pe:
            print(f"PermissionError : {pe}")
except FileNotFoundError as fnf:
    print(f"FileNotFoundError : {fnf}")
else:
    print(f"Data in file : {data}")
finally:
    print("File operation attempted")
```

### Task - 5 - Mini Program : Safe Shopping Cart


```python
cart = []
while True:
    menuinput = input("Menu : \n"
                      "1. Enter Price : \n"
                      "q: Press q to quit")
    if menuinput == "1":
        inputPrice = input("Enter the price : ")
        try:
            inputPrice = float(inputPrice)
            if not isinstance(inputPrice,(int,float)):
                raise ValueError("Invalid Input")
            elif float(inputPrice) < 0:
                raise Exception("Negative not allowed")

        except ValueError as ve:
            print(f"ValueError : {ve}")
        except Exception as e:
            print(f"Exception : {e}")
        else:
            cart.append(inputPrice)

    elif menuinput == "q":
        break
    else:
        continue

print(f"Total Bills : {sum(cart)}")
print(f"Total Itmes : {len(cart)}")


```
