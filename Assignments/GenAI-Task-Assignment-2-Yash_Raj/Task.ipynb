### Task 1 : Discount Rules (if /elif / else)


```python
order_amount = input("Enter the order amount : ")

if order_amount.isdigit():
    order_amount = int(order_amount)
    if order_amount >= 2000:
        finalAmount = order_amount - (order_amount * (15/100))
    elif order_amount >= 1500 and order_amount < 2000:
        finalAmount = order_amount - (order_amount * (10/100))
    elif order_amount >= 1000 and order_amount <=1500:
        finalAmount = order_amount - (order_amount * (7/100))
    else:
        finalAmount = order_amount
        print(f"Discount is Zero")

    amount_after_tax = finalAmount + (finalAmount * (5/100))
    print(f"Order Amount : {order_amount}\n"
          f"Sub Total Amount after discount : {finalAmount}\n"
          f"Tax on discount : {(finalAmount * (5/100))}\n"
          f"Final Total Amount after Tax : {amount_after_tax}")

else:
    print("Invalid type")
```

### Task 2 : Process Multiple Orders (for Loop)


```python
orders = [1200,2500,800,1750,3000]
ordersDiscount_count = 0
revenueAfterDiscount = 0
for order_amount in orders:
    if order_amount >= 2000:
        finalAmount = order_amount - (order_amount * (15/100))
        ordersDiscount_count +=1

    elif order_amount >= 1500 and order_amount < 2000:
        finalAmount = order_amount - (order_amount * (10/100))
        ordersDiscount_count +=1
    elif order_amount >= 1000 and order_amount <=1500:
        finalAmount = order_amount - (order_amount * (7/100))
        ordersDiscount_count +=1
    else:
        finalAmount = order_amount
        print(f"Discount is Zero")

    amount_after_tax = finalAmount + (finalAmount * (5/100))
    revenueAfterDiscount +=amount_after_tax
    print(f"Order Amount : {order_amount} -> "
           f"Discount % {((order_amount-finalAmount)/order_amount)*100} -> "
          f"Final Total Amount after Tax 5%: {amount_after_tax}\n")
print(f"Revenue after discount : {revenueAfterDiscount}")
print(f"Numbers of orders that recieved discount : {ordersDiscount_count}")
```

### Task 3: User Menu (while loop + break/continue)


```python
order_list = []

while True:
    menu_input = input(f"Menu:\n"
          f"1. Add order amount: \n"
          f"2. Show all orders and total after discounts \n"
          f"q. Quit \n"
          f"Selct Options from Menu (1/2/q)")

    if menu_input == "1":
        order = int(input("Add order amount: "))
        order_list.append(order)
    elif menu_input == "2":
        for order_amount in order_list:
            if order_amount >= 2000:
                finalAmount = order_amount - (order_amount * (15/100))
                ordersDiscount_count +=1

            elif order_amount >= 1500 and order_amount < 2000:
                finalAmount = order_amount - (order_amount * (10/100))
                ordersDiscount_count +=1
            elif order_amount >= 1000 and order_amount <=1500:
                finalAmount = order_amount - (order_amount * (7/100))
                ordersDiscount_count +=1
            else:
                finalAmount = order_amount
                print(f"Discount is Zero")

            amount_after_tax = finalAmount + (finalAmount * (5/100))
            revenueAfterDiscount +=amount_after_tax
            print(f"Order Amount : {order_amount} -> "
                   f"Discount % {((order_amount-finalAmount)/order_amount)*100} -> "
                  f"Final Total Amount after Tax 5%: {amount_after_tax}\n")
    elif menu_input == "q":
        break
    else:
        continue
```

### Task 5 : Loop Control with Conditions (break & Continue)


```python
daily =  [200,150,0,400,-1,300]
total_sales = 0
for item in daily:
    if item == -1:
        print("Data corrupted, Stopping execution")
        break
    elif item == 0:
        print(f"Sales are 0 today, No sales")
        continue
    else:
        total_sales +=item

print(f"Total Sales : {total_sales}")

```
