## Task 1 : Pandas Series Basics


```python
import pandas as pd

marks = [78,85,90,66,72]
marks_df = pd.Series(name="Marks", data=marks)
print(f"Series Values : {marks_df}")

print(f"Values Index : {marks_df.index}")

print(f"DataType : {marks_df.dtypes}")

print(f"Accessing First Element : {marks_df[0]}")

print(f"Accessing Last  two elements : {marks_df[3:]}")
```

### Task 2 : Mathematical Operations on Series


```python
#add 5 grace marls to all students
print(f"Adding 5 grace marks to all students : {marks_df + 5}")

#Subtract 2 marks from all values
print(f"Subtract 2 marks from all values : {marks_df - 2}")

#Multiply all marks by 1.05
print(f"Multiply all marks by 1.05: {marks_df * 1.05}")

#Devide all marks by 2
print(f"Devide all marks by 2: {marks_df / 2}")
```

### Task 3 : Python functionalities on Series



```python
#Finding maximum marks
print(f"Finding maximum marks : {marks_df.max()}")

#Finding minimum marks
print(f"Finding minimum marks : {marks_df.min()}")

#Sum of marks
print(f"Sum of marks : {marks_df.sum()}")

#Mean of marks
print(f"Mean of marks : {marks_df.mean()}")

#lambda function check whether each student has passed (>=70)
newList = [marks_df[x] for x in range(len(marks_df)) if marks_df.values[x] >=70]
print(f"Passed Students with number : {newList}")

#count how many students passed
print(f"count how many students passed : {len(newList)}")

```

### Task 4 : Create a DataFrame



```python
#Create dataframe students

students = {
    'Name' : ['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks' : [78,85,90,66,72],
    'Subject' : ['Math','Math','Science','Science','Math']
}

students_df = pd.DataFrame(students)
print(f"Student Dataframe :\n {students_df}")

#Printing first 3 rows
print(f"Printing first 3 rows : {students_df.head(3)}")

#Printing last 3 rows
print(f"Printing last 3 rows : {students_df.tail(2)}")

#Printing dataframe shape and column name
print(f"Printing dataframe shape : {students_df.shape} and column name : {students_df.columns}")
```

### Task 5: Important DataFrame Functions


```python
students = {
    'Name' : ['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks' : [78,85,90,66,72],
    'Subject' : ['Math','Math','Science','Science','Math']
}
students_df = pd.DataFrame(students)
#.info()
print(f"Info : {students_df.info()}")

#.describe()
print(f"Describe : {students_df.describe()}")

#.head()
print(f"Head : {students_df.head()}")

#.tail()
print(f"Tail : {students_df.tail()}")

#Sort students by marks in descending order
print(f"Sort students by marks in descending order : {students_df.sort_values(by='Marks',ascending=False)}")

#reset index
print(f"Reset index: {students_df.sort_values(by='Marks',ascending=False).reset_index()}")
```

### Task 6: Filtering and conditional Selection



```python
students = {
    'Name' : ['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks' : [78,85,90,66,72],
    'Subject' : ['Math','Math','Science','Science','Math']
}
students_df = pd.DataFrame(students)

#Students who scored that 75 marks
print(f"Students who scored that 75 marks : \n {students_df[students_df["Marks"] > 75]}")

#Students belongs to subject Math
print(f"Students belongs to subject Math : \n {students_df[students_df["Subject"] == "Math"]}")

#Students who scored than average marks
print(f"Students who scored than average marks : \n  {students_df[students_df["Marks"] > students_df["Marks"].mean()]}")

#Students who failed (marks < 70)
print(f"Students who failed (marks < 70) : \n {students_df[students_df["Marks"] < 75]}")
```

### Task 7: Grouping and basic Analysis



```python
students = {
    'Name' : ['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks' : [78,85,90,66,72],
    'Subject' : ['Math','Math','Science','Science','Math']
}
students_df = pd.DataFrame(students)

#Find Average marks per subject using groupby()
print(f"Average marks per subject : \n {students_df.groupby('Subject')['Marks'].mean()}")

#Count number of students per subject
print(f"Count number of students per subject : \n {students_df['Subject'].value_counts()}")

#Maximun Marks per subject
print(f"Maximun Marks per subject : \n {students_df.groupby('Subject')['Marks'].max()}")
```

### Task 8 : Pandas Plotting ( Simple Graphs )


```python
students = {
    'Name' : ['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks' : [78,85,90,66,72],
    'Subject' : ['Math','Math','Science','Science','Math']
}
students_df = pd.DataFrame(students)

#Plot a bar graph of student name vs marks
sorted_students = students_df.sort_values('Marks',ascending=False)
print(sorted_students.plot(kind='bar', x='Name', y='Marks'))

#Plot line graph for marks
print(sorted_students.plot(kind='bar'))

#Plot histogram for marks
print(sorted_students.plot(kind='hist'))
```

```python
# Create the DataFrame
sales = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}
df = pd.DataFrame(sales)

print("--- Sales Data Analysis Results ---")

# 1. Total revenue
total_revenue = df['Revenue'].sum()
print(f"1. Total Revenue: {total_revenue}")

# 2. Average daily revenue
avg_revenue = df['Revenue'].mean()
print(f"2. Average Daily Revenue: {avg_revenue}")

# 3. Day with highest revenue
# idxmax() finds the index row, and .loc extracts the 'Day' from it
highest_revenue_day = df.loc[df['Revenue'].idxmax(), 'Day']
print(f"3. Day with Highest Revenue: {highest_revenue_day}")

# 4. Days where revenue > average
# We filter the DataFrame using a conditional condition
high_revenue_days = df[df['Revenue'] > avg_revenue]['Day'].tolist()
print(f"4. Days where revenue > average: {high_revenue_days}")

# 5. Plot revenue vs day
df.plot(kind='bar', x='Day', y='Revenue', color='teal', edgecolor='black', legend=False)
```
