## Task 1  - Product Collection (List & Tuples)



```python
from decorator import append


def productCollection(products, sample_products, new_products):
    print(f"Second Product from list : {products[1]}")
    print(f"Last product from list : {products[-1]}")

    for product in new_products:
        products.append(product)

    print(f"Updated Product List : {products}")

    sample_products = list(sample_products)
    sample_products[1] = "600"
    print(f"Converted Sample prodcust back to tupple after price change : {tuple(sample_products)}")



products = ["ToyCar","Perfume","Bell","Tyres","CarHood"]
sample_products = ("CarSeat","590","Cars")
new_prodcuts = ["KeyChain","Brush"]
productCollection(products,sample_products,new_prodcuts)
```

## Task 2 - Categories (Sets)


```python
def categories(categories_list):
    categories_set = set(categories_list)
    print(f"Products Categories list converted to set : {categories_set}")

    categories_set.add("ToyCar")
    print(f"After adding 'ToyCar' as category, it will get ignored : {categories_set}")
    print(f"Length of unique item in set : {len(categories_set)}")
    if "ToyCar" in categories_set:
        return True
    else:
        return False


categories_list = ["ToyCar","Perfume","Bell","Tyres","CarHood"]

existflag = categories(categories_list)
print(f"Item Exist! : {existflag}")
```

## Task 3 - Product Pricing (Dictionary)


```python
price_dict = {"ToyCar":120,
              "Perfume":540.25,
              "CarHood":6000,
              "Bell":460,
              "Tyres":3000,
              "TravelBag":1500
              }

print(f"Product and price dict : {price_dict}")

price_dict["SleepingBag"] = 999
print(f"Product and price dict after adding new product : {price_dict}")

price_dict["Bell"] = 200
print(f"Product and price dict after updating price for Bell product : {price_dict}")

del price_dict["Bell"]
print(f"Product and price dict after deleting Bell product : {price_dict}")

sum = 0
for value in price_dict.values():
    sum+=value
print(f"Average price of all products : {sum/len(price_dict)}")

maxx = int(price_dict["ToyCar"])
maxProduct = "ToyCar"
minn = int(price_dict["ToyCar"])
minProduct = "ToyCar"


for key in price_dict.keys():
    if price_dict.get(key) > maxx:
        maxx = price_dict.get(key)
        maxProduct = key

for key1 in price_dict.keys():
    if price_dict.get(key1) < minn:
        minn = price_dict.get(key1)
        minProduct = key1

print(f"Max Value Prodcut : {maxProduct}, Price : {maxx}")
print(f"Min Value Prodcut : {minProduct},  Price : {minn}")

```
